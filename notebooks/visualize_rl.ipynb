```python
import json
import matplotlib.pyplot as plt

import os
import sys

# Add the project root to the Python path to allow imports from the agents module
sys.path.append("..")
os.environ["ROOT_DIR"] = ".."  # this allows to use relative paths in config files
```

```python
with open('../logs/rl/square_corridor/curriculum/static/training_stats.json') as f:
    log = json.load(f)

keys = log[0].keys()

print(keys)
for key in keys:
    if key not in ['epoch', 'rew_pen', 'cum_rew_ext']:
        lst = []
        for line in log:
            lst.append(line[key])
        plt.plot(lst)
        plt.grid()
        plt.xlabel('epoch')
        plt.title(key)
        plt.show()
```
