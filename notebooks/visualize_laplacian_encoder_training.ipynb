```python
import json
import matplotlib.pyplot as plt

maze_type = 'square_corridor'
exp_name = 'curriculum'
use_curriculum = True
stage = 2
if use_curriculum:
    with open('../logs/laplacian_encoder/{0}/{1}/stage_{2}/training_stats.json'.format(maze_type, exp_name, stage), encoding='utf-8') as f:
        data = json.load(f)
else:
    with open('../logs/laplacian_encoder/{0}/{1}/training_stats.json'.format(maze_type, exp_name), encoding='utf-8') as f:
        data = json.load(f)

key_lst = data[0].keys()
for key in key_lst:
    lst = []
    for l in data:
        lst.append(l[key])

    plt.title(key)
    plt.plot(lst)
    plt.show()
```
